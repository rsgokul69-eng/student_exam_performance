```python
df =df.drop(columns="student_id")
```

#splitting data

```python
x = df.drop("pass_status", axis=1)
y = df["pass_status"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2,random_state=42,stratify=y)
```

```python
# Splitting into numerical and categorical columns

numerical_cols = x_train.select_dtypes(include="number").columns

categorical_cols = x_train.select_dtypes(include="object").columns

print("Numerical columns:", numerical_cols)
print("Categorical columns:", categorical_cols)
```

#numerical data preprocessing

```python
#handling missing values -numerical columns
num_imputer=SimpleImputer(strategy="median")

x_train_num=num_imputer.fit_transform(x_train[numerical_cols])
x_test_num=num_imputer.transform(x_test[numerical_cols])
```

#scaling

```python
scaler = StandardScaler()

x_train_num_scale = scaler.fit_transform(x_train_num)
x_test_num_scale = scaler.transform(x_test_num)
```

```python
cat_imputer=SimpleImputer(strategy="most_frequent")

x_train_cat=cat_imputer.fit_transform(x_train[categorical_cols])
x_test_cat=cat_imputer.transform(x_test[categorical_cols])
```

```python
for col in categorical_cols:
  print(f"{col}:===>{df[col].unique()}")
```

```python
onehot_cols = [
    "gender",
    "education_level",
    "school_type",
    "parent_education",
    "urban_rural",
    "study_environment",
    "study_method",
    "device_availability"
]

ordinal_cols = [
    "family_income",
    "class_participation",
    "study_consistency",
    "revision_frequency",
    "notes_quality",
    "sleep_quality",
    "break_frequency",
    "motivation_level",
    "educational_app_usage",
    "exam_difficulty"
]
```

```python
x_train_cat_df = pd.DataFrame(
    x_train_cat,
    columns=categorical_cols,
    index=x_train.index
)

x_test_cat_df = pd.DataFrame(
    x_test_cat,
    columns=categorical_cols,
    index=x_test.index
)
```

```python
from sklearn.preprocessing import OneHotEncoder
onehot_encoder = OneHotEncoder(handle_unknown="ignore")

x_train_onehot = onehot_encoder.fit_transform(
    x_train_cat_df[onehot_cols]
)

x_test_onehot = onehot_encoder.transform(
    x_test_cat_df[onehot_cols]
)

print(x_train_onehot.shape)
```

```python
from sklearn.preprocessing import OrdinalEncoder

ordinal_cols = [
    "family_income",
    "class_participation",
    "study_consistency",
    "revision_frequency",
    "notes_quality",
    "sleep_quality",
    "break_frequency",
    "motivation_level",
    "educational_app_usage",
    "exam_difficulty"
]

ordinal_cols = [col for col in ordinal_cols if col in categorical_cols]

ordinal_categories = [
    ["Low", "Lower-Middle", "Middle", "Upper-Middle", "High"],
    ["Low", "Medium", "High"],
    ["Low", "Medium", "High"],
    ["Rarely", "Weekly", "Daily"],
    ["Poor", "Average", "Excellent"],
    ["Poor", "Fair", "Good", "Excellent"],
    ["Rarely", "Occasionally", "Frequently"],
    ["Low", "Medium", "High"],
    ["Low", "Moderate", "High"],
    ["Easy", "Medium", "Hard"]
]

# Keep only categories corresponding to columns present
ordinal_categories = [
    ordinal_categories[
        ["family_income", "class_participation", "study_consistency",
         "revision_frequency", "notes_quality", "sleep_quality",
         "break_frequency", "motivation_level", "educational_app_usage",
         "exam_difficulty"].index(col)
    ]
    for col in ordinal_cols
]

ordinal_encoder = OrdinalEncoder(
    categories=ordinal_categories,
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

x_train_ordinal = ordinal_encoder.fit_transform(
    x_train_cat_df[ordinal_cols]
)

x_test_ordinal = ordinal_encoder.transform(
    x_test_cat_df[ordinal_cols]
)

print(x_train_ordinal)
```

```python
# One hot encoded dataframe
onehot_df = pd.DataFrame(
    x_train_onehot.toarray(),
    columns=onehot_encoder.get_feature_names_out(onehot_cols),
    index=x_train.index
)

# Ordinal encoded dataframe
ordinal_df = pd.DataFrame(
    x_train_ordinal,
    columns=ordinal_cols,
    index=x_train.index
)

print(onehot_df.head())
print(ordinal_df.head())
```

```python
encoded_cat_df = pd.concat(
    [onehot_df, ordinal_df], axis=1
)

print(encoded_cat_df.head())
```

```python
# Convert numerical array into DataFrame
numerical_df = pd.DataFrame(
    x_train_num_scale,
    columns=numerical_cols,
    index=x_train.index
)

# Combine numerical and categorical data
x_train_final = pd.concat(
    [numerical_df, encoded_cat_df], axis=1
)

print(x_train_final.head())
print(x_train_final.shape)
```

```python
# One hot test dataframe
onehot_test_df = pd.DataFrame(
    x_test_onehot.toarray(),
    columns=onehot_encoder.get_feature_names_out(onehot_cols),
    index=x_test.index
)

# Ordinal test dataframe
ordinal_test_df = pd.DataFrame(
    x_test_ordinal,
    columns=ordinal_cols,
    index=x_test.index
)

# Combine test categorical data
encoded_test_cat_df = pd.concat(
    [onehot_test_df, ordinal_test_df], axis=1
)

# Combine test numerical and categorical data
numerical_test_df = pd.DataFrame(
    x_test_num_scale,
    columns=numerical_cols,
    index=x_test.index
)

x_test_final = pd.concat(
    [numerical_test_df, encoded_test_cat_df],
    axis=1
)

print(x_test_final.head())
print(x_test_final.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

print("Target classes:", label_encoder.classes_)
print(y_train_encoded)
```

```python
from imblearn.over_sampling import SMOTE

# Check class distribution before SMOTE
print(y_train.value_counts())

# Apply SMOTE
smote = SMOTE(random_state=42)

x_train_balanced, y_train_balanced = smote.fit_resample(
    x_train_final, y_train_encoded
)

# Check class distribution after SMOTE
print(pd.Series(y_train_balanced).value_counts())
```
